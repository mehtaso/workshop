```sql
-- Drop schema and all associated objects if they exist
drop schema if exists workshop cascade;

-- Create a new schema/database called "workshop"
create schema workshop;
```

```sql
%python

# Read data from the csv file and store in a Spark DataFrame
diamonds = spark.read.csv("/databricks-datasets/Rdatasets/data-001/csv/ggplot2/diamonds.csv", header="true", inferSchema="true")

# Write the data to a table called workshop.diamonds in Delta format, overwrite if the table already exists
diamonds.write.format("delta").mode("overwrite").saveAsTable("workshop.diamonds")
```

```sql
-- This command provides detailed metadata, including the table's schema, file system information, owner, and access time

describe table extended workshop.diamonds;
```

```sql
-- View count of the table
select count(*) from workshop.diamonds;
```

```sql
-- Let's look at the data
select * from workshop.diamonds;
```

```sql
-- Let's look at the history of the table to see what commands have been run against it
DESCRIBE HISTORY workshop.diamonds;
```

```sql
-- Let's delete some data
DELETE FROM workshop.diamonds WHERE cut == 'Good';
```

```sql
-- View count of the table
select count(*) from workshop.diamonds;
```

```sql
-- Let's look at the history of the table again to see if anything has changed
DESCRIBE HISTORY workshop.diamonds;
```

```sql
-- Let's look at version 0 and confirm that all of the data is still there
SELECT * FROM workshop.diamonds VERSION AS OF 0;
```

```sql
-- Let's undo the delete command and restore version 0
RESTORE TABLE workshop.diamonds TO VERSION AS OF 0;
```

```sql
-- Let's run a count on the table to confirm that the restore command worked
select count(*) from workshop.diamonds;
```

```sql
-- Let's run some SQL commands against the table
SELECT color, avg(price) AS price FROM workshop.diamonds GROUP BY color ORDER BY COLOR
```
